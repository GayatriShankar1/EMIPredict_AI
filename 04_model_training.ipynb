```python
# Classification Model Training
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

# 1. Load preprocessed dataset
try:
    df = pd.read_csv('data/processed_emi_dataset.csv')
except FileNotFoundError:
    df = pd.read_csv('/Users/gayatrishankar/Downloads/emi_prediction_dataset.csv', low_memory=False)

# 2. Define features (X) and target (y)
X = df.drop(columns=['emi_eligibility', 'max_monthly_emi'], errors='ignore')
y_class = df['emi_eligibility']

# 3. Train-Test Split (80% train, 20% test)
X_train, X_test, y_train_c, y_test_c = train_test_split(
    X, y_class, test_size=0.2, random_state=42
)

# 4. Train Classification Model
clf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
clf.fit(X_train, y_train_c)

# 5. Evaluate Model
y_pred_c = clf.predict(X_test)

print("=== CLASSIFICATION REPORT (EMI Eligibility) ===")
print(classification_report(y_test_c, y_pred_c))
print(f"Accuracy Score: {accuracy_score(y_test_c, y_pred_c) * 100:.2f}%")
```

```python
# Train Regression Model (max_monthly_emi)
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# 1. Define target for regression
y_reg = df['max_monthly_emi']

# 2. Train-Test Split for Regression
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X, y_reg, test_size=0.2, random_state=42
)

# 3. Train Regression Model
reg = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
reg.fit(X_train_r, y_train_r)

# 4. Evaluate Model
y_pred_r = reg.predict(X_test_r)

mae = mean_absolute_error(y_test_r, y_pred_r)
rmse = np.sqrt(mean_squared_error(y_test_r, y_pred_r))
r2 = r2_score(y_test_r, y_pred_r)

print("=== REGRESSION REPORT (Max Monthly EMI) ===")
print(f"Mean Absolute Error (MAE): ₹{mae:.2f}")
print(f"Root Mean Squared Error (RMSE): ₹{rmse:.2f}")
print(f"R² Score: {r2 * 100:.2f}%")
```

```python
# Save Trained Models & Encoders
import joblib
import os

# Create models directory if it doesn't exist
os.makedirs('models', exist_ok=True)

# Save Models
joblib.dump(clf, 'models/emi_classifier.pkl')
joblib.dump(reg, 'models/emi_regressor.pkl')

print("Classification & Regression models have been saved to the 'models/' directory!")
```
